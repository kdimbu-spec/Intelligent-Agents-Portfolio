```python
import nltk
from nltk import Tree

tree_str = """(S
   (NP
      (Det The)
      (N government))
   (VP
      (V raised)
      (NP
         (N interest)
         (N rates))))"""

tree = Tree.fromstring(tree_str)
print(tree)
tree.pretty_print()
```
