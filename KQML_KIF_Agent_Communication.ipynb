### Create the Warehouse Agent (Bob)
#### KQML Message Processing

```python
class Bob:

    def __init__(self):
        self.stock = {
            "television50inch": 120
        }

        self.specifications = {
            "television50inch": {
                "hdmi_slots": 4
            }
        }

    def process_message(self, message):

        performative = message["performative"]

        if performative == "ask-stock":

            tv = message["content"]

            stock_level = self.stock.get(tv, 0)

            return {
                "performative": "tell",
                "sender": "Bob",
                "receiver": "Alice",
                "content": f"(stock-level {tv} {stock_level})"
            }

        elif performative == "ask-hdmi":

            tv = message["content"]

            hdmi = self.specifications[tv]["hdmi_slots"]

            return {
                "performative": "tell",
                "sender": "Bob",
                "receiver": "Alice",
                "content": f"(hdmi-slots {tv} {hdmi})"
            }
```

### Create the Procurement Agent (Alice)

```python
class Alice:

    def ask_stock(self):

        return {
            "performative": "ask-stock",
            "sender": "Alice",
            "receiver": "Bob",
            "content": "television50inch"
        }

    def ask_hdmi_slots(self):

        return {
            "performative": "ask-hdmi",
            "sender": "Alice",
            "receiver": "Bob",
            "content": "television50inch"
        }
```

```python
alice = Alice()
bob = Bob()
```

### Alice Sends the First KQML Query

```python
message = alice.ask_stock()

print("KQML Message Sent:")
print(message)
```

### Bob Processes the Query

```python
response = bob.process_message(message)

print("KQML Response:")
print(response)
```

### Alice Requests HDMI Information

```python
message2 = alice.ask_hdmi_slots()

print("KQML Message Sent:")
print(message2)
```

### Bob Replies with the HDMI Info

```python
response2 = bob.process_message(message2)

print("KQML Response:")
print(response2)
```

### The Complete Dialogue

```python
print("===== AGENT DIALOGUE =====")

msg1 = alice.ask_stock()
print("\nAlice -> Bob")
print(msg1)

reply1 = bob.process_message(msg1)
print("\nBob -> Alice")
print(reply1)

msg2 = alice.ask_hdmi_slots()
print("\nAlice -> Bob")
print(msg2)

reply2 = bob.process_message(msg2)
print("\nBob -> Alice")
print(reply2)
```
